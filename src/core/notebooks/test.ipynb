```python
import pdfplumber

def count_words_in_pdf(pdf_path):
    total_words = 0
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                # split() handles any whitespace (spaces, tabs, newlines)
                words = text.split()
                total_words += len(words)
    return total_words

# Usage
path = "../data/raw/EMV_v4.4_Book_1_ICC_to_Terminal_Interface.pdf"
print(f"Total Words: {count_words_in_pdf(path)}")
```
